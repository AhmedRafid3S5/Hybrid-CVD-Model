```python
import tensorflow as tf
from tensorflow import keras
import numpy as np

# Load the model
model = keras.models.load_model("simple_4_model.h5")

# Print model summary
model.summary()

# Access weights for each layer
for i, layer in enumerate(model.layers):
    print(f"\n--- Layer {i}: {layer.name} ({layer.__class__.__name__}) ---")
    weights = layer.get_weights()
    
    if weights:
        for j, w in enumerate(weights):
            print(f"Weight matrix {j} shape: {w.shape}")
            print(f"Weight matrix {j}:\n{w}")
    else:
        print("No weights in this layer")

# Access specific layer weights by index or name
first_layer_weights = model.layers[0].get_weights()
print(f"\nFirst layer weights: {first_layer_weights}")

# Get weights as a list of numpy arrays for all layers
all_weights = model.get_weights()
print(f"\nTotal number of weight arrays: {len(all_weights)}")
```
